# 🧪 Prueba de la librería `algebra-lineal-sheets`

Este notebook verifica que la librería instalada en el entorno `.venv` funciona de punta a punta usando el **modo Excel local** (sin Google ni credenciales):

1. Exporta las matrices `A` y `b` al archivo `mis_matrices.xlsx` (una pestaña por matriz).
2. Lista las pestañas con `workspace()`.
3. Vuelve a importar `A` y `b` desde el Excel.
4. Resuelve el sistema lineal $Ax = b$ y exporta la solución `x`.
5. Comprueba el contenido del archivo generado.

> **Kernel:** selecciona el intérprete de `.venv` (Python 3.12) al ejecutar este notebook en VS Code.

```python
import numpy as np
from algebra_lineal import *

version()
```

```python
# Modo Excel local: no requiere internet ni cuenta de Google
configurar(sheet='mis_matrices.xlsx')
```

```python
# Crear las matrices del sistema y exportarlas al Excel
A = np.array([[3.0, 1.0],
              [1.0, 2.0]])
b = np.array([[9.0],
              [8.0]])

exportar('A', 'b')
```

```python
# Ver las pestañas del archivo y sus dimensiones
workspace()
```

```python
# Borrar las variables del namespace y volver a importarlas desde el Excel
del A, b
importar('A', 'b')

print('A =')
print(A)
print('b =')
print(b)
```

```python
# Resolver Ax = b, verificar la solución y exportarla
x = np.linalg.solve(A, b)
print('x =')
print(x)

assert np.allclose(A @ x, b), 'La solución no verifica A @ x = b'
print('✅ Verificado: A @ x ≈ b')

exportar('x')
```

```python
# Comprobar el archivo Excel generado leyéndolo directamente con openpyxl
from pathlib import Path
from openpyxl import load_workbook

ruta = Path('mis_matrices.xlsx')
print('Archivo existe:', ruta.exists())
print('Ubicación:', ruta.resolve())

libro = load_workbook(ruta, data_only=True)
print('Pestañas:', libro.sheetnames)
print('Pestaña x:', [[celda.value for celda in fila] for fila in libro['x'].iter_rows()])
```
